DATA LOADING & READING
# Railway Delay Analyzer

```python
import pandas as pd
import numpy as np
from google.colab import files
```

Upload Dataset

```python
uploaded = files.upload()

file_name = list(uploaded.keys())[0]

print("Dataset loaded successfully!")
print("File Name:", file_name)
```

 Read Dataset

```python
df = pd.read_csv(file_name)
```

Dataset Shape

```python
print("\n" + "=" * 50)
print("DATASET BASIC INFORMATION")
print("=" * 50)

print("Number of Rows    :", df.shape[0])
print("Number of Columns :", df.shape[1])
```

Column Names

```python
print("\n" + "=" * 50)
print("COLUMN NAMES")
print("=" * 50)

for i, column in enumerate(df.columns, start=1):
    print(i, ".", column)
```

```python
print("\n" + "=" * 50)
print("FIRST 5 RECORDS")
print("=" * 50)

display(df.head())
```

Last 5 Records

```python
print("\n" + "=" * 50)
print("LAST 5 RECORDS")
print("=" * 50)

display(df.tail())
```

```python
print("\n" + "=" * 50)
print("DATA TYPES")
print("=" * 50)

display(df.dtypes.to_frame(name="Data Type"))
```

Save Raw Dataset

```python
df.to_csv("stage1_raw_data.csv", index=False)

print("\n" + "=" * 50)
print("STAGE 1 COMPLETED")
print("=" * 50)

print("Output file created: stage1_raw_data.csv")
```
