DATA ACQUISITION & FILTERING
# Railway Delay Analyzer

```python
import pandas as pd
from google.colab import files
```

Upload Dataset

```python
uploaded = files.upload()

file_name = list(uploaded.keys())[0]

print("Dataset acquired successfully!")
print("File Name:", file_name)
```

Read Dataset

```python
df = pd.read_csv(file_name)

print("\n" + "=" * 55)
print("DATASET ACQUISITION")
print("=" * 55)

print("Rows    :", df.shape[0])
print("Columns :", df.shape[1])
```

Check Required Columns

```python
required_columns = [
    "date",
    "train_no",
    "train_name",
    "type_code",
    "station_no",
    "station_name",
    "delay"
]

missing_columns = [
    column for column in required_columns
    if column not in df.columns
]

print("\n" + "=" * 55)
print("REQUIRED COLUMN VALIDATION")
print("=" * 55)

if len(missing_columns) == 0:
    print("All required columns are available.")
else:
    print("Missing columns:", missing_columns)
```

Boolean Indexing — Delayed Trains

```python
delayed_trains = df[df["delay"].fillna(0) > 0].copy()

print("\n" + "=" * 55)
print("BOOLEAN FILTERING")
print("=" * 55)

print("Total records              :", len(df))
print("Delayed records (> 0 min)  :", len(delayed_trains))
```

filter() — Select Required Columns

```python
delayed_report = delayed_trains.filter(
    items=[
        "date",
        "train_no",
        "train_name",
        "station_name",
        "delay"
    ]
)

print("\n" + "=" * 55)
print("FILTERED DELAYED TRAIN DATA")
print("=" * 55)

display(delayed_report.head(10))
```

Filter Delays Above 15 Minutes

```python
delay_over_15 = df[df["delay"].fillna(0) > 15].copy()

print("\n" + "=" * 55)
print("SEVERE DELAY FILTER")
print("=" * 55)

print("Records with delay > 15 minutes:",
      len(delay_over_15))
```

Display Severe Delay Records

```python
severe_delay_report = delay_over_15.filter(
    items=[
        "date",
        "train_no",
        "train_name",
        "station_name",
        "delay"
    ]
)

display(severe_delay_report.head(10))
```

Save Stage 2 Outputs

```python
delayed_report.to_csv(
    "stage2_delayed_trains.csv",
    index=False
)

severe_delay_report.to_csv(
    "stage2_delays_over_15_minutes.csv",
    index=False
)
```

Completion Message

```python
print("\n" + "=" * 55)
print("STAGE 2 COMPLETED")
print("=" * 55)

print("Output 1: stage2_delayed_trains.csv")
print("Output 2: stage2_delays_over_15_minutes.csv")
```
