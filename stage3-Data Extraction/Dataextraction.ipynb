Upload Dataset & Read

```python
import pandas as pd
from google.colab import files

# Upload dataset
uploaded = files.upload()

file_name = list(uploaded.keys())[0]

# Read CSV
df = pd.read_csv(file_name)

print("Dataset loaded successfully!")
print("Original Shape:", df.shape)

df.head()
```

Convert Date Columns

```python
# Convert date column
df["date"] = pd.to_datetime(df["date"], errors="coerce")

# Convert datetime columns
df["scheduled_arrival_datetime"] = pd.to_datetime(
    df["scheduled_arrival_datetime"], errors="coerce"
)

df["actual_arrival_datetime"] = pd.to_datetime(
    df["actual_arrival_datetime"], errors="coerce"
)

print("Date columns converted successfully!")

print("\nData Types:")
print(df[[
    "date",
    "scheduled_arrival_datetime",
    "actual_arrival_datetime"
]].dtypes)
```

Extract Date Information

```python


df["extracted_year"] = df["date"].dt.year
df["extracted_month"] = df["date"].dt.month
df["extracted_day"] = df["date"].dt.day
df["extracted_day_of_week"] = df["date"].dt.day_name()

print("Date information extracted successfully!")

print(df[[
    "date",
    "extracted_year",
    "extracted_month",
    "extracted_day",
    "extracted_day_of_week"
]].head())
```

Select / Extract Important Columns

```python


extracted_columns = [
    "date",
    "train_no",
    "train_name",
    "type_code",
    "station_no",
    "station_name",
    "station_full_name",
    "station_zone",
    "distance_from_origin",
    "arrival_day",
    "arrival_time",
    "departure_day",
    "departure_time",
    "delay",
    "year",
    "month",
    "day_of_week",
    "scheduled_arrival_minutes",
    "scheduled_departure_minutes",
    "scheduled_arrival_datetime",
    "actual_arrival_datetime",
    "arrival_delay_minutes",
    "scheduled_vs_actual_arrival_status",
    "weather_condition",
    "temperature_c",
    "rainfall_mm",
    "congestion_index",
    "congestion_level",
    "operational_disruption",
    "extracted_year",
    "extracted_month",
    "extracted_day",
    "extracted_day_of_week"
]

extracted_df = df[extracted_columns].copy()

print("Data extraction completed!")
print("Extracted Shape:", extracted_df.shape)

extracted_df.head()
```

Extract Delayed Train Records

```python


delayed_trains = extracted_df[
    extracted_df["delay"].fillna(0) > 0
].copy()

print("Delayed train records extracted successfully!")
print("Number of delayed records:", len(delayed_trains))

delayed_trains[[
    "date",
    "train_no",
    "train_name",
    "station_name",
    "delay"
]].head()
```

Extract Weather & Operational Information

```python
weather_operational_data = extracted_df[[
    "date",
    "train_no",
    "station_name",
    "delay",
    "weather_condition",
    "temperature_c",
    "rainfall_mm",
    "congestion_index",
    "congestion_level",
    "operational_disruption"
]].copy()

print("Weather and operational data extracted!")
print("Shape:", weather_operational_data.shape)

weather_operational_data.head()
```

Save Extracted Data

```python

extracted_df.to_csv(
    "stage3_extracted_data.csv",
    index=False
)

# Save delayed records
delayed_trains.to_csv(
    "stage3_delayed_train_records.csv",
    index=False
)

# Save weather and operational data
weather_operational_data.to_csv(
    "stage3_weather_operational_data.csv",
    index=False
)

print("Files saved successfully!")
print("\nCreated files:")
print("1. stage3_extracted_data.csv")
print("2. stage3_delayed_train_records.csv")
print("3. stage3_weather_operational_data.csv")
```
