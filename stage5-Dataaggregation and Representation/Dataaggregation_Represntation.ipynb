Upload Cleaned Dataset

```python
import pandas as pd
from google.colab import files

# Upload Stage 4 cleaned dataset
uploaded = files.upload()

file_name = list(uploaded.keys())[0]

df = pd.read_csv(file_name)

# Convert date column
df["date"] = pd.to_datetime(df["date"], errors="coerce")

print("Stage 4 cleaned dataset loaded successfully!")
print("Dataset Shape:", df.shape)
```

Basic Dataset Validation

```python
print("STAGE 5 DATASET VALIDATION")
print("=" * 60)

print("Rows:", df.shape[0])
print("Columns:", df.shape[1])

print("Unique Trains:", df["train_no"].nunique())
print("Unique Stations:", df["station_no"].nunique())

print("Start Date:", df["date"].min())
print("End Date:", df["date"].max())
```

Train Performance using groupby() + agg()

Idi mandatory requirement.

```python
train_performance = (
    df.groupby(
        ["train_no", "train_name"]
    )
    .agg(
        total_records=("train_no", "size"),
        average_delay=("delay", "mean"),
        maximum_delay=("delay", "max"),
        minimum_delay=("delay", "min"),
        total_delay=("delay", "sum")
    )
    .reset_index()
)

print("TRAIN PERFORMANCE")
print("=" * 60)

display(train_performance.head(10))
```

Station Performance using groupby() + agg()

```python
station_performance = (
    df.groupby(
        ["station_no", "station_name"]
    )
    .agg(
        total_records=("station_no", "size"),
        average_delay=("delay", "mean"),
        maximum_delay=("delay", "max"),
        minimum_delay=("delay", "min"),
        total_delay=("delay", "sum")
    )
    .reset_index()
)

print("STATION PERFORMANCE")
print("=" * 60)

display(station_performance.head(10))
```

Delayed Trains using Boolean Indexing

```python
delayed_trains = df[
    df["delay"] > 0
].copy()

print("DELAYED TRAIN RECORDS")
print("=" * 60)

print("Total delayed records:", len(delayed_trains))

display(
    delayed_trains[
        [
            "date",
            "train_no",
            "train_name",
            "station_name",
            "delay"
        ]
    ].head(10)
)
```

Delay > 15 Minutes using Boolean Indexing + filter()

```python
high_delay = df[
    df["delay"] > 15
].copy()

# Select required columns using filter()
high_delay_selected = high_delay.filter(
    items=[
        "date",
        "train_no",
        "train_name",
        "station_name",
        "delay",
        "weather_condition",
        "congestion_level",
        "operational_disruption"
    ]
)

print("HIGH DELAY RECORDS")
print("=" * 60)

print(
    "Records with delay > 15 minutes:",
    len(high_delay_selected)
)

display(high_delay_selected.head(10))
```

Train Ranking using sort_values()

```python
train_ranking = (
    train_performance
    .sort_values(
        by="average_delay",
        ascending=False
    )
    .reset_index(drop=True)
)

print("TRAIN RANKING BY AVERAGE DELAY")
print("=" * 60)

display(train_ranking.head(10))
```

Station Ranking using sort_values()

```python
station_ranking = (
    station_performance
    .sort_values(
        by="average_delay",
        ascending=False
    )
    .reset_index(drop=True)
)

print("STATION RANKING BY AVERAGE DELAY")
print("=" * 60)

display(station_ranking.head(10))
```

merge() Train Details with Performance

```python
train_details = (
    df[
        [
            "train_no",
            "train_name",
            "type_code"
        ]
    ]
    .drop_duplicates()
)

train_summary = train_performance.merge(
    train_details,
    on=[
        "train_no",
        "train_name"
    ],
    how="left"
)

print("MERGE OPERATION")
print("=" * 60)

print("Train performance and train details merged successfully!")

display(train_summary.head(10))
```

Create Delay Categories

```python
df["delay_category"] = pd.cut(
    df["delay"],
    bins=[
        -float("inf"),
        0,
        15,
        float("inf")
    ],
    labels=[
        "On Time",
        "Delay 1-15 Minutes",
        "Delay >15 Minutes"
    ]
)

print("Delay categories created successfully!")

print(
    df["delay_category"].value_counts()
)
```

pivot_table() Station-wise Delay Comparison

```python
station_delay_pivot = pd.pivot_table(
    df,
    index="station_name",
    columns="delay_category",
    values="delay",
    aggfunc="count",
    fill_value=0,
    observed=False
)

print("STATION-WISE DELAY COMPARISON")
print("=" * 60)

display(
    station_delay_pivot.head(10)
)
```

transform() for Train-level Comparison

Idi mandatory requirement.

```python
df["train_average_delay"] = (
    df.groupby("train_no")["delay"]
    .transform("mean")
)

df["delay_vs_train_average"] = (
    df["delay"] -
    df["train_average_delay"]
)

print("TRAIN-LEVEL TRANSFORM COMPARISON")
print("=" * 60)

display(
    df[
        [
            "train_no",
            "train_name",
            "station_name",
            "delay",
            "train_average_delay",
            "delay_vs_train_average"
        ]
    ].head(10)
)
```

transform() for Station-level Comparison

```python
df["station_average_delay"] = (
    df.groupby("station_no")["delay"]
    .transform("mean")
)

df["delay_vs_station_average"] = (
    df["delay"] -
    df["station_average_delay"]
)

print("STATION-LEVEL TRANSFORM COMPARISON")
print("=" * 60)

display(
    df[
        [
            "station_no",
            "station_name",
            "delay",
            "station_average_delay",
            "delay_vs_station_average"
        ]
    ].head(10)
)
```

Save Stage 5 Outputs

```python
# Train performance
train_performance.to_csv(
    "stage5_train_performance.csv",
    index=False
)

# Station performance
station_performance.to_csv(
    "stage5_station_performance.csv",
    index=False
)

# Delayed trains
delayed_trains.to_csv(
    "stage5_delayed_trains.csv",
    index=False
)

# High delay records
high_delay_selected.to_csv(
    "stage5_high_delay_records.csv",
    index=False
)

# Merged train summary
train_summary.to_csv(
    "stage5_train_summary_merged.csv",
    index=False
)

# Pivot table
station_delay_pivot.to_csv(
    "stage5_station_delay_pivot.csv"
)

# Final aggregated dataset
df.to_csv(
    "stage5_aggregated_dataset.csv",
    index=False
)

print("All Stage 5 files saved successfully!")

print("\nFiles Created:")
print("1. stage5_train_performance.csv")
print("2. stage5_station_performance.csv")
print("3. stage5_delayed_trains.csv")
print("4. stage5_high_delay_records.csv")
print("5. stage5_train_summary_merged.csv")
print("6. stage5_station_delay_pivot.csv")
print("7. stage5_aggregated_dataset.csv")
```
